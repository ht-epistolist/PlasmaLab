# Modal Comparison: Acquired Waveforms vs Controller PWM

This notebook demonstrates overlaying acquired scope waveforms (from a selected data folder) with the controller PWM signal (calculated using the mathematical expressions from `coil_test.ipynb`).

Both plots share the same x-axis (Time in ms), with the PWM waveform aligned to start at the detected oscilloscope shot trigger time (default: 10 ms).

```python
from modal_compare import modal_compare
from IPython.display import Image, display
import os
```

## Example 1: Sawtooth Waveform Comparison

Overlay the acquired `sawtooth1.csv` raw and filtered signals with a simulated sawtooth controller PWM signal.

```python
modal_compare(
    data_dir='compare_data',
    filename='sawtooth1.csv',
    pwm_type='sawtooth',
    channels=['CH4'],
    CH_ref='CH4',
    yunit='A',
    colormap='plasma',
    plot_denoised=True,
    output_filename='modal_example_sawtooth.png'
)

display(Image(filename='plotes_color/modal_example_sawtooth.png'))
```
